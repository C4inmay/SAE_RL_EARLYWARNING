# ECG and HRV Feature Extraction

**Project:** RL-Based Early Warning of Neurological Deterioration in Sepsis ICU Patients

## Objective

Link the sepsis ICU cohort with MIMIC-IV waveform records and derive ECG-based Heart Rate Variability (HRV) features.

### Pipeline

ECG waveform
→ signal preprocessing
→ R-peak detection
→ RR intervals
→ HRV features
→ hourly aggregation

```python
from pathlib import Path

import numpy as np
import pandas as pd

PROJECT_ROOT = Path("..")

PROCESSED_DIR = (
    PROJECT_ROOT /
    "data" /
    "processed"
)

WAVEFORM_DIR = (
    PROJECT_ROOT /
    "data" /
    "raw" /
    "mimic-waveform"
)

WAVEFORM_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Processed:", PROCESSED_DIR)
print("Waveform:", WAVEFORM_DIR)
```

```python
cohort = pd.read_csv(
    PROCESSED_DIR /
    "sepsis_icu_cohort_demo.csv"
)

subject_ids = sorted(
    cohort["subject_id"]
    .astype(int)
    .unique()
)

print("Patients:", len(subject_ids))
print(subject_ids)
```
